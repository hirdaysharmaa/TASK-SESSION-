### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
first_term=int(input("enter 1cterm"))
second_term=int(input ("enter 2term "))
n=int(input("enter nth term "))
d=second_term-first_term
nth_term=first_term+(n-1)*d
print(nth_term)
````

Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
import math
h_t=float(input("h"))
b_t=float(input("b"))
l_t=float(input("l"))

h_g=float(input("hight of glass"))
r_g=float(input("radius of glass"))

volume_tank=h_t *b_t *l_t
volume_glass=3.14*r_g*r_g*h_g
print('no of glasses',math.floor(volume_tank/volume_glass))
````
